```python
from typing import reveal_type
#normal function
def div(a,b):
  print(a/b)

# adding decorators
def magic_div(fun):

  def inner(a,b): #inner() is the wrapper that adds new behavior

   if a<b:
     a,b=b,a
   return fun(a,b) # Call the original function with potentially swapped arguments
  return inner # magic_div must return the inner function

div2=magic_div(div)
div2(2,4)
```

```python
#normal function
def greet(name):
    return f"hello, {name}!"

# decorator to convert result to uppercase
def uppercase_decorator(func):
    def inner(name):
        result=func(name)
        return result.upper()
    return inner

# apply decorator manually
greet_upper=uppercase_decorator(greet)

#call the decorated function
print(greet_upper("alice"))
```
