```python
import pandas as pd

sales = pd.Series(
    [1200,850,1500,950,200],
    index=["monday","tuesday","wednesday","thursday","friday"]
    )

print(sales)
print(sales.max())
print(sales.sum())
print(sales.min())
print(sales[sales<1000])
print(sales[sales>1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)    #series was used for one line number
print(ar)
print(ar.unique()) #nique was used for removing repeated values
print(ar.nunique()) #nunique was used for show how many repeated number
print(ar.value_counts()) #value_count is used for  the data how much value of numbers have there
```

```python
data=[20,30,None,None,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.isnull()) #isnull --- is show true on null value and show false on not null

print(ar.notnull()) #notnull --- is show true on not null and show false on null value
```
